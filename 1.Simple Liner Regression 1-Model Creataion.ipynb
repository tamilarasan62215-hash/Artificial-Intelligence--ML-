```python
import pandas as pd               # import pandas library using as pd
import matplotlib.pyplot as plt   # import matplotlibrary access an pyplot as plt 
```

```python
dataset=pd.read_csv("Salary_Data.csv")  # dataset variable store in pandas(lib) in comma saprated value stored (DATAS ip & op)
```

```python
dataset                            # output for dataseet  
```

```python
independent=dataset[["YearsExperience"]]  # store the independent valu[[input "YearsExperience"]] data x
independent
```

```python
dependent=dataset[["Salary"]]    # store the dependent valu [[output "Salary"]] data y
dependent
```

```python
plt.scatter(independent,dependent)      # use the matplotlibrary as (plt)to creat graph and find this a simple liner
plt.xlabel('yearsExperience',fontsize=20) # use a xlabel for x-valu in size of 20
plt.ylabel('Salary',fontsize=20)          # use a ylabel for y-valu in size of 20
plt.show()                                # output for graph
```

```python
from sklearn.model_selection import train_test_split  # python(lib) in model selection to split train & test 
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)# giving an parameter value x,y train test
x_train,x_test,y_train,y_test
```

```python
from sklearn.linear_model import LinearRegression # py(lib)in linear model from linear gression algorithm formula 
regressor=LinearRegression()            # regressor store all data or model formula
regressor.fit(x_train,y_train)          # fit key use process to learn from x & y train 
```

```python
weight=regressor.coef_        # coef key find the weight(range) between two point 
weight
```

```python
bais=regressor.intercept_    # intercept key find the starting point 
bais
```

```python
y_pred=regressor.predict(x_test)     # predict key process x_test pridition y_predition vallu store in y_pred 
y_pred
```

```python
from sklearn.metrics import r2_score   # r2 checking how model it's working with actual value to predicicted value
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import pickle           # pickle to save a model
filename="finalized_model_linear.sav"
```

```python
pickle.dump(regressor,open(filename,'wb')) # pickle in dump regressor(valus) in new file with (writte binary)
```

```python
loaded_model=pickle.load(open("finalized_model_linear.sav",'rb')) # process to file in read binary 
result=loaded_model.predict([[13]])                               # input for predict experience staff 
result
```
